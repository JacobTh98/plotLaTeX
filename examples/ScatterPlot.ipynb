# Scatterplot

```python
import matplotlib.pyplot as plt
import numpy as np

from plotLaTeX import ScatterPlot

np.random.seed(seed=42)
```

```python
# create random data
x = np.linspace(0, 10, 50)
y = np.sin(x) + np.random.normal(scale=0.2, size=len(x))
```

```python
plt.figure(figsize=(7.5, 3))
plt.scatter(x, y, label="Data points")
plt.xlabel("x-label")
plt.ylabel("x-label")
plt.grid(alpha=0.3)
plt.show()
```

```python
Splt = ScatterPlot()
```

```python
# add the x reference
Splt.add_xvals(x)
```

```python
# add the first y signal
Splt.add_yvals(y, y_name="y")
```

```python
# save the data as ".csv" and get the latex figure code
Splt.export(f_name="scatter_results.csv")
```
